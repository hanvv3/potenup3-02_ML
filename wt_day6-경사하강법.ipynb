# 경사하강법

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.random.rand(100)
y = 0.2*X
```

```python
plt.figure(figsize=(6,4))
plt.scatter(X, y)
```

```python
w = np.random.uniform(-1,1)
b = np.random.uniform(-1,1)
w
```

```python
pred = w*X+b

def plot_pred(y, pred):
    plt.figure(figsize=(6,4))
    plt.scatter(X, y)
    plt.scatter(X, pred)
    plt.show
```

```python
plot_pred(y, pred)
```
